```python
df = pd.read_csv('netflix_titles.csv')
df =df.drop_duplicates()
df.corr(numeric_only=True)
```

```python
plt.figure(figsize=(12, 6))
plt.plot(df["release_year"], linewidth=2)
plt.xlabel("Index", fontsize=12)
plt.ylabel("Release Year", fontsize=12)
plt.title("Release Year Trend", fontsize=14)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 8))
top_countries = df["country"].value_counts().head(10)
plt.pie(top_countries, labels=top_countries.index, autopct='%1.1f%%', startangle=90)
plt.title("Top 10 Countries by Netflix Content", fontsize=14)
plt.tight_layout()
plt.show()
```

```python
plt.hist(df["release_year"])
plt.xlabel("Release Year", fontsize=12)
plt.ylabel("Frequency", fontsize=12)
plt.title("Distribution of Release Years", fontsize=14)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))
plt.scatter(df["release_year"], df.index, alpha=0.5, s=20)
plt.xlabel("Release Year", fontsize=12)
plt.ylabel("Content Index", fontsize=12)
plt.title("Content Distribution by Release Year", fontsize=14)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
df.boxplot(column="release_year")
plt.title("Boxplot of Release Years", fontsize=14)
plt.tight_layout()
plt.show()
```
